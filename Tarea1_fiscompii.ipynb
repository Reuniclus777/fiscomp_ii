```python

def calcule_energia_cinetica(m, vmkh):
    v_ms = vmkh * (1000 / 3600)
    e_k = 0.5*m*(v_ms**2)
    return e_k
masa = 4
velocidad = 2
energia = calcule_energia_cinetica(masa, velocidad)
print(f"la energia cinetica es : {energia} joules")
```
